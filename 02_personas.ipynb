# Component 2 — Synthetic Personas

We define 5 synthetic personas representing people living with dementia, each with distinct preferences over image features. Each persona has:
- A written profile (background, age)
- A quantitative preference vector (weights over image features)

The personas are designed to be distinct enough for a bandit to learn but realistic for the TalkStory/Moodshifter use case.

```python
import json

with open("personas.json") as f:
    personas = json.load(f)["personas"]

for p in personas:
    print(f"\n{'='*60}")
    print(f"{p['name']}, age {p['age']}")
    print(f"{p['background']}")
    print(f"\nPreference weights:")
    for feature, weight in sorted(p['preferences'].items(), key=lambda x: -x[1]):
        bar = '+' * int(abs(weight) * 10) if weight >= 0 else '-' * int(abs(weight) * 10)
        print(f"  {feature:25s} {weight:+.1f}  {bar}")
```

## Persona Design Rationale

| Persona | Key positive features | Key negative features | Why distinct |
|---------|----------------------|----------------------|-------------|
| Eleanor | plants, outdoor, children | dim, indoor | Nature-loving grandmother |
| Frank | water, cool tones, cityscape | indoor, plants | Urban waterfront worker |
| Margaret | plants, warm, outdoor | crowds (num_people), cool | Quiet gardener |
| James | people, bright, adults | dim | Social/active coach |
| Dorothy | indoor, warm, evening | crowds, bright, cool | Quiet introverted librarian |

Each persona occupies a different region of the feature preference space, ensuring the bandit must learn genuinely different policies for each.
